```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
img = cv2.imread('img.jpeg')
```

```python
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
plt.imshow(img_rgb)
plt.title("Original Image")
plt.axis("off")
plt.show()
```

```python
#1. Smoothing filter
#i) Using Averaging Filter
```

```python
averaging = cv2.blur(img, (5, 5))
plt.imshow(cv2.cvtColor(averaging, cv2.COLOR_BGR2RGB))
plt.title("Averaging Filter")
plt.axis("off")
plt.show()
```

```python
#ii) Using Weighted Averaging Filter (custom kernel)
```

```python
kernel = np.array([[1,2,1],[2,4,2],[1,2,1]]) / 16
weighted = cv2.filter2D(img, -1, kernel)
plt.imshow(cv2.cvtColor(weighted, cv2.COLOR_BGR2RGB))
plt.title("Weighted Averaging")
plt.axis("off")
plt.show()
```

```python
#iii) Using Gaussian Filter
```

```python

gaussian = cv2.GaussianBlur(img, (5, 5), 0)
plt.imshow(cv2.cvtColor(gaussian, cv2.COLOR_BGR2RGB))
plt.title("Gaussian Filter")
plt.axis("off")
plt.show()
```

```python
#iv) Using Median Filter
```

```python

median = cv2.medianBlur(img, 5)
plt.imshow(cv2.cvtColor(median, cv2.COLOR_BGR2RGB))
plt.title("Median Filter")
plt.axis("off")
plt.show()
```

```python
#2. Sharpening Filters
#i) Using Laplacian Kernel (Manual Kernel)
```

```python
kernel = np.array([[0,-1,0],[-1,5,-1],[0,-1,0]])
laplacian_sharp = cv2.filter2D(img, -1, kernel)

plt.imshow(cv2.cvtColor(laplacian_sharp, cv2.COLOR_BGR2RGB))
plt.title("Laplacian Sharpening")
plt.axis("off")
plt.show()
```

```python
#ii) Using Laplacian Operator (OpenCV built-in)
```

```python
gray = cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
laplacian = cv2.Laplacian(gray, cv2.CV_64F)
laplacian = cv2.convertScaleAbs(laplacian)
plt.imshow(laplacian, cmap='gray')
plt.title("Laplacian Operator")
plt.axis("off")
plt.show()
```

```python
images = [img_rgb, averaging, weighted, gaussian, median, laplacian_sharp, laplacian]
titles = ["Original", "Averaging", "Weighted", "Gaussian", "Median", "Laplacian Kernel", "Laplacian Operator"]
plt.figure(figsize=(12, 8))
for i in range(7):
    plt.subplot(3, 3, i+1)
    plt.imshow(images[i] if i == 6 else cv2.cvtColor(images[i], cv2.COLOR_BGR2RGB), cmap='gray' if i == 6 else None)
    plt.title(titles[i])
    plt.axis('off')
plt.tight_layout()
plt.show()
```
